## 실습 2. 전체승차인원(유임+무임)이 만명이상인 역 중, 유임승차 비율이 가장 높은 역은 어디입니까?

```python
df['전체승차인원'] = df['유임승차'] + df['무임승차']
```

```python
df.loc[ df['전체승차인원'] >= 10000 , ].sort_values('비율', ascending=False)
```
